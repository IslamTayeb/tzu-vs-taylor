## 1. Text → numbers
A pre-trained AI reads each line and squeezes its **meaning** into 384 numbers, called an **embedding**.

```
#@title
tvt.text_to_numbers("Trash takes itself out every single time.")
```

Lines that mean similar things get similar numbers, even if they share no words.

```
#@title
tvt.closest("Trash takes itself out every single time.")
```

## 2. Train
The model learns which patterns in those numbers mean Taylor and which mean Sun Tzu, then we test it on lines it has **never seen**.

```
#@title
tvt.train()
```

## 4. What did it learn?
We feed it single words, one at a time, and see which ones it's most sure about.

```
#@title
tvt.top_words(15)
```

And here is every line as a dot. The ★ stars are our trap lines.

```
#@title
tvt.vector_map()
```

**Notice:** it learned *topics*: war words go to Sun Tzu, love words go to Taylor. So a line about war in Taylor's voice can fool it, and so can Sun Tzu in Gen Z slang.
