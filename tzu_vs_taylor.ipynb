## 4. What did it actually learn?
The words that push hardest toward each answer.

```
#@title
tvt.top_words(15)
```

**Notice:** Sun Tzu gets *enemy, army, ground* — but also boring words like *of, is, the*. Taylor gets *you, me, my*. It mostly learned **who is talking**, not **what about**. That's a shortcut — and shortcuts can be fooled.
