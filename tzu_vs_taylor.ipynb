## 1. The data
Every row is one line and who said it.

```
from tvt import load, embed, chance_of_taylor, show

lines = load()          # equal numbers of Sun Tzu and Taylor lines
lines.sample(5)
```

## 2. Text to numbers
A pre-trained AI model reads each line and turns its meaning into 384 numbers, called an *embedding*.

```
vectors = embed(lines.text)   # one row of 384 numbers per line
vectors.shape
```

```
show.stripe("Trash takes itself out every single time.")
```

Lines that mean similar things get similar numbers, even with no words in common.

```
show.closest(lines, vectors, "My ex is spreading rumors about me")
```

## 3. Train, then test
Hide 20% of the lines, learn from the rest, then check the hidden ones.

```
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

x_train, x_test, y_train, y_test = train_test_split(
    vectors, lines.label, test_size=0.2, stratify=lines.label, random_state=0)

model = LogisticRegression(max_iter=2000).fit(x_train, y_train)   # learn which patterns mean Taylor
show.accuracy(model.score(x_test, y_test), len(y_test))           # grade it on lines it never saw
```
